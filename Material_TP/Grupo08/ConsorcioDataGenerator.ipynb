```python
import pandas as pd
import random
```

```python
df = pd.read_excel('../Miel/Archivos-para-el-TP/Archivos para el TP/datos varios.xlsx', sheet_name='Consorcios')
df.head()
```

```python
df['nro'] = df['Domicilio'].str.extract(r'(\d+)$').astype(float).astype('Int64')
df['calle'] = df['Domicilio'].str.replace(r'\s*\d+\s*$', '', regex=True)
df.head()
```

```python
df['localidad'] = 'Ciudad Autónoma de Buenos Aires'
df['provincia'] = 'Ciudad Autónoma de Buenos Aires'
df['cuit'] = [f"30-{random.randint(10000000, 99999999)}-{random.randint(0,9)}" for _ in range(len(df))]
df['razon_social'] = df['Nombre del consorcio'] + " S.A."
bancos = ['Banco Nación', 'Banco Galicia', 'Banco BBVA', 'Banco Santander']
df['banco'] = [random.choice(bancos) for _ in range(len(df))]
df['cbu_cvu'] = [f"{random.randint(1000000000000000000000, 9999999999999999999999)}"[:22] for _ in range(len(df))]
df.head()
```

```python
df.to_csv('./data/consorcios.csv', index=False, encoding='utf-8')
```
